Our real target project is Gigaphoton-style AI-assisted laser beam alignment. Future real data may include BP/BD beam images, COG, size, symmetry, ContM, uniformity, DTLs, HV, Energy Sigma, Dose/HK, F2 pressure, wavelength, Twin Evaluation, session history, and calibration metadata.

```python
import csv
import pandas as pd
from pathlib import Path

print("File import successful")
```

```python
DATASET_PATH = Path('../data/laser_dataset')
IMAGES_PATH  = DATASET_PATH / 'images'
assert DATASET_PATH.exists(), f'Dataset not found at {DATASET_PATH}'
print(f'Dataset found: {DATASET_PATH}')
# Proposed Path
OUTPUT_CSV = DATASET_PATH / 'labels/custom_labels_v1.csv'
OUTPUT_CSV.parent.mkdir(parents=True, exist_ok=True)

# The file will be created later when you write something to it.
df = pd.DataFrame([
    {
        "sample_id": "sample_0001",
        "image_path": "images/beam_001.png",
        "beam_quality_label": "UNKNOWN"
    }
])

# This line creates the actual CSV file
df.to_csv(OUTPUT_CSV, index=False)

assert OUTPUT_CSV.exists(), f'Directory not found: {OUTPUT_CSV}'
print(f'Output Directory Created Successfully: {OUTPUT_CSV}')
```

| Column | Meaning |
|---|---|
| `sample_id` | Unique row ID like `sample_0001` |
| `image_path` | Relative path to the PNG |
| `source_dataset` | Where it came from, e.g. `laser_resonator_subset_v1` |
| `beam_image_type_proxy` | Whether this is BP-like or BD-like image; we will guess later |
| `beam_quality_label` | Overall verdict: `GOOD`, `SHIFTED`, `DISTORTED`, etc. |
| `alignment_state_label` | Specific failure mode, if any |
| `severity_label` | `PASS`, `WARN`, or `FAIL` gate for safety logic |
| `giga_rank_proxy` | Maps to `S/A/B/C/D/E` in the main project |
| `center_status` | Is the beam centered? |
| `size_status` | Is the beam the right size? |
| `symmetry_status` | Is the beam symmetric? |
| `uniformity_status` | Is intensity spread evenly? |
| `contm_proxy_status` | Is contour shape normal? Analog of `ContM` |
| `notes` | Free-text comments per row |
|

```python
# core label columns (will be filled later)
CORE_COLUMNS = [
    "sample_id", "image_path", "source_dataset",
    "beam_image_type_proxy", "beam_quality_label", "alignment_state_label",
    "severity_label", "giga_rank_proxy",
    "center_status", "size_status", "symmetry_status",
    "uniformity_status", "contm_proxy_status",
    "notes",
]

# 1. list all PNGs
image_files = sorted(IMAGES_PATH.glob("*.png"))
print(f"Found {len(image_files)} images")

# 2. write skeleton CSV
with open(OUTPUT_CSV, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=CORE_COLUMNS)
    writer.writeheader()

    for i, img_path in enumerate(image_files, start=1):
        row = {col: "UNKNOWN" for col in CORE_COLUMNS}
        row["sample_id"]      = f"sample_{i:04d}"
        row["image_path"]     = str(img_path.relative_to(IMAGES_PATH.parent))
        row["source_dataset"] = "laser_resonator_subset_v1"
        row["notes"]          = ""
        writer.writerow(row)

print(f"CSV created at {OUTPUT_CSV}")
```
